```python
# ============================================================
# A* SEARCH ALGORITHM FOR 8-PUZZLE
# Heuristic: Manhattan Distance
# g(n) = Depth
# h(n) = Manhattan Distance
# f(n) = g(n) + h(n)
# ============================================================

import heapq

# ------------------------------------------------------------
# INITIAL STATE AND GOAL STATE
# ------------------------------------------------------------

initial_state = (
    2, 8, 3,
    1, 6, 4,
    7, 5, 0
)

goal_state = (
    1, 2, 3,
    8, 0, 4,
    7, 6, 5
)

# 0 represents the blank space


# ------------------------------------------------------------
# DISPLAY PUZZLE
# ------------------------------------------------------------

def print_puzzle(state):

    for i in range(0, 9, 3):
        print(state[i], state[i + 1], state[i + 2])

    print()


# ------------------------------------------------------------
# HEURISTIC: MANHATTAN DISTANCE
# ------------------------------------------------------------

def manhattan_distance(state, goal):

    distance = 0

    for tile in range(1, 9):

        # Current position of tile
        current_position = state.index(tile)

        # Goal position of tile
        goal_position = goal.index(tile)

        # Current row and column
        current_row = current_position // 3
        current_col = current_position % 3

        # Goal row and column
        goal_row = goal_position // 3
        goal_col = goal_position % 3

        # Manhattan distance
        distance += abs(current_row - goal_row)
        distance += abs(current_col - goal_col)

    return distance


# ------------------------------------------------------------
# FIND POSSIBLE MOVES
# ------------------------------------------------------------

def get_neighbors(state):

    neighbors = []

    blank_position = state.index(0)

    row = blank_position // 3
    col = blank_position % 3

    # Possible blank movements
    moves = [
        (-1, 0),   # Up
        (1, 0),    # Down
        (0, -1),   # Left
        (0, 1)     # Right
    ]

    for dr, dc in moves:

        new_row = row + dr
        new_col = col + dc

        # Check valid position
        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_position = new_row * 3 + new_col

            new_state = list(state)

            # Swap blank with tile
            new_state[blank_position], new_state[new_position] = \
                new_state[new_position], new_state[blank_position]

            neighbors.append(tuple(new_state))

    return neighbors


# ------------------------------------------------------------
# RECONSTRUCT SOLUTION PATH
# ------------------------------------------------------------

def reconstruct_path(parent, goal):

    path = []

    current = goal

    while current is not None:

        path.append(current)

        current = parent[current]

    path.reverse()

    return path


# ------------------------------------------------------------
# A* SEARCH
# ------------------------------------------------------------

def a_star_manhattan(initial, goal):

    # Priority queue
    open_list = []

    counter = 0

    # g(n) values
    g_cost = {
        initial: 0
    }

    # Parent of each state
    parent = {
        initial: None
    }

    # Calculate initial heuristic
    h = manhattan_distance(initial, goal)

    # f(n) = g(n) + h(n)
    f = 0 + h

    heapq.heappush(
        open_list,
        (f, counter, initial)
    )

    # CLOSED set
    closed = set()

    print("A* SEARCH - MANHATTAN DISTANCE")
    print("=" * 50)

    while open_list:

        # Select state having minimum f
        f_current, _, current = heapq.heappop(open_list)

        # Ignore already expanded states
        if current in closed:
            continue

        # Put current state in CLOSED
        closed.add(current)

        current_g = g_cost[current]
        current_h = manhattan_distance(current, goal)
        current_f = current_g + current_h

        print("Current State:")
        print_puzzle(current)

        print("g(n) =", current_g)
        print("h(n) =", current_h)
        print("f(n) =", current_f)
        print("-" * 50)

        # ----------------------------------------------------
        # GOAL TEST
        # ----------------------------------------------------

        if current == goal:

            print("GOAL STATE REACHED!")
            print()

            return reconstruct_path(parent, current)

        # ----------------------------------------------------
        # GENERATE SUCCESSORS
        # ----------------------------------------------------

        for neighbor in get_neighbors(current):

            if neighbor in closed:
                continue

            # Every move has cost 1
            new_g = current_g + 1

            # New state or cheaper path
            if neighbor not in g_cost or new_g < g_cost[neighbor]:

                g_cost[neighbor] = new_g

                parent[neighbor] = current

                # Manhattan heuristic
                neighbor_h = manhattan_distance(
                    neighbor,
                    goal
                )

                # f(n) = g(n) + h(n)
                neighbor_f = new_g + neighbor_h

                counter += 1

                heapq.heappush(
                    open_list,
                    (
                        neighbor_f,
                        counter,
                        neighbor
                    )
                )

    return None


# ------------------------------------------------------------
# RUN A*
# ------------------------------------------------------------

solution = a_star_manhattan(
    initial_state,
    goal_state
)


# ------------------------------------------------------------
# PRINT FINAL SOLUTION PATH
# ------------------------------------------------------------

if solution:

    print("\n")
    print("=" * 50)
    print("FINAL SOLUTION PATH")
    print("=" * 50)

    print("Number of moves:", len(solution) - 1)
    print()

    for step, state in enumerate(solution):

        print("Step", step)
        print_puzzle(state)

else:

    print("No solution found.")
```

```python
# ============================================================
# A* SEARCH ALGORITHM FOR 8-PUZZLE
# Heuristic: Number of Misplaced Tiles
# g(n) = Depth
# h(n) = Number of Misplaced Tiles
# f(n) = g(n) + h(n)
# ============================================================

import heapq

# ------------------------------------------------------------
# INITIAL STATE AND GOAL STATE
# ------------------------------------------------------------

initial_state = (
    2, 8, 3,
    1, 6, 4,
    7, 5, 0
)

goal_state = (
    1, 2, 3,
    8, 0, 4,
    7, 6, 5
)

# 0 represents the blank space


# ------------------------------------------------------------
# DISPLAY PUZZLE
# ------------------------------------------------------------

def print_puzzle(state):
    for i in range(0, 9, 3):
        print(state[i], state[i + 1], state[i + 2])
    print()


# ------------------------------------------------------------
# HEURISTIC: MISPLACED TILES
# ------------------------------------------------------------

def misplaced_tiles(state, goal):
    count = 0

    for i in range(9):
        # Do not count blank (0)
        if state[i] != 0 and state[i] != goal[i]:
            count += 1

    return count


# ------------------------------------------------------------
# FIND POSSIBLE MOVES
# ------------------------------------------------------------

def get_neighbors(state):

    neighbors = []

    blank_position = state.index(0)

    row = blank_position // 3
    col = blank_position % 3

    # Possible movements of blank
    moves = [
        (-1, 0),   # Up
        (1, 0),    # Down
        (0, -1),   # Left
        (0, 1)     # Right
    ]

    for dr, dc in moves:

        new_row = row + dr
        new_col = col + dc

        # Check whether new position is inside 3x3 board
        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_position = new_row * 3 + new_col

            new_state = list(state)

            # Swap blank with adjacent tile
            new_state[blank_position], new_state[new_position] = \
                new_state[new_position], new_state[blank_position]

            neighbors.append(tuple(new_state))

    return neighbors


# ------------------------------------------------------------
# A* SEARCH
# ------------------------------------------------------------

def a_star_misplaced(initial, goal):

    # Priority queue
    # Stores: (f, counter, state)
    open_list = []

    counter = 0

    # g(initial) = 0
    g_cost = {initial: 0}

    # Parent dictionary for reconstructing path
    parent = {initial: None}

    # Calculate h and f
    h = misplaced_tiles(initial, goal)

    f = 0 + h

    heapq.heappush(open_list, (f, counter, initial))

    # CLOSED set
    closed = set()

    print("A* SEARCH - MISPLACED TILES")
    print("=" * 50)

    while open_list:

        # Select state with minimum f
        f_current, _, current = heapq.heappop(open_list)

        # Skip if already expanded
        if current in closed:
            continue

        # Add to CLOSED
        closed.add(current)

        current_g = g_cost[current]
        current_h = misplaced_tiles(current, goal)
        current_f = current_g + current_h

        print("Current State:")
        print_puzzle(current)

        print("g(n) =", current_g)
        print("h(n) =", current_h)
        print("f(n) =", current_f)
        print("-" * 50)

        # ----------------------------------------------------
        # GOAL TEST
        # ----------------------------------------------------

        if current == goal:

            print("GOAL STATE REACHED!")
            print()

            return reconstruct_path(parent, current)

        # ----------------------------------------------------
        # GENERATE SUCCESSORS
        # ----------------------------------------------------

        for neighbor in get_neighbors(current):

            if neighbor in closed:
                continue

            new_g = current_g + 1

            # If this is a new state OR we found a cheaper path
            if neighbor not in g_cost or new_g < g_cost[neighbor]:

                g_cost[neighbor] = new_g
                parent[neighbor] = current

                neighbor_h = misplaced_tiles(neighbor, goal)
                neighbor_f = new_g + neighbor_h

                counter += 1

                heapq.heappush(
                    open_list,
                    (neighbor_f, counter, neighbor)
                )

    return None


# ------------------------------------------------------------
# RECONSTRUCT SOLUTION PATH
# ------------------------------------------------------------

def reconstruct_path(parent, goal):

    path = []

    current = goal

    while current is not None:
        path.append(current)
        current = parent[current]

    path.reverse()

    return path


# ------------------------------------------------------------
# RUN A*
# ------------------------------------------------------------

solution = a_star_misplaced(initial_state, goal_state)


# ------------------------------------------------------------
# PRINT FINAL SOLUTION PATH
# ------------------------------------------------------------

if solution:

    print("\n")
    print("=" * 50)
    print("FINAL SOLUTION PATH")
    print("=" * 50)

    print("Number of moves:", len(solution) - 1)
    print()

    for step, state in enumerate(solution):

        print("Step", step)
        print_puzzle(state)

else:

    print("No solution found.")
```
