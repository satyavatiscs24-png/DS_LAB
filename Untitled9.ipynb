```python
# 8-Puzzle Problem using Depth First Search (DFS)

from collections import deque

# Initial state
initial = (
    1, 2, 3,
    4, 5, 6,
    7, 0, 8
)

# Goal state
goal = (
    1, 2, 3,
    4, 5, 6,
    7, 8, 0
)


# Function to print the puzzle
def print_puzzle(state):
    for i in range(0, 9, 3):
        print(state[i], state[i+1], state[i+2])
    print()


# Generate possible next states
def get_neighbors(state):

    neighbors = []

    # Find position of blank (0)
    zero_position = state.index(0)

    row = zero_position // 3
    col = zero_position % 3

    # Possible movements: Up, Down, Left, Right
    moves = [
        (-1, 0, "UP"),
        (1, 0, "DOWN"),
        (0, -1, "LEFT"),
        (0, 1, "RIGHT")
    ]

    for dr, dc, move in moves:

        new_row = row + dr
        new_col = col + dc

        # Check whether the move is valid
        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_position = new_row * 3 + new_col

            # Create new state
            new_state = list(state)

            # Swap blank with the selected tile
            new_state[zero_position], new_state[new_position] = \
                new_state[new_position], new_state[zero_position]

            neighbors.append((tuple(new_state), move))

    return neighbors


# DFS function
def dfs(initial, goal):

    # Stack stores (state, path)
    stack = [(initial, [])]

    # Visited states
    visited = set()

    while stack:

        # Remove the last element (LIFO)
        current, path = stack.pop()

        # Skip if already visited
        if current in visited:
            continue

        visited.add(current)

        # Check goal
        if current == goal:
            return path, visited

        # Generate next states
        neighbors = get_neighbors(current)

        # Add neighbors to stack
        for next_state, move in neighbors:

            if next_state not in visited:
                stack.append((next_state, path + [move]))

    return None, visited


# Run DFS
solution, visited = dfs(initial, goal)


# Display result
print("INITIAL STATE:")
print_puzzle(initial)

print("GOAL STATE:")
print_puzzle(goal)

if solution is not None:

    print("Goal Found!")
    print("Number of moves:", len(solution))
    print("Moves:", solution)

else:

    print("No solution found.")

print("Number of states visited:", len(visited))
```

```python
# 8-Puzzle using Iterative Deepening Search (IDS)

# Initial state
initial = (
    1, 2, 3,
    4, 5, 6,
    0, 7, 8
)

# Goal state
goal = (
    1, 2, 3,
    4, 5, 6,
    7, 8, 0
)


# Function to print puzzle
def print_puzzle(state):
    for i in range(0, 9, 3):
        print(state[i], state[i+1], state[i+2])
    print()


# Generate possible moves
def get_neighbors(state):

    neighbors = []

    # Find blank position
    zero = state.index(0)

    row = zero // 3
    col = zero % 3

    # UP, DOWN, LEFT, RIGHT
    moves = [
        (-1, 0, "UP"),
        (1, 0, "DOWN"),
        (0, -1, "LEFT"),
        (0, 1, "RIGHT")
    ]

    for dr, dc, move in moves:

        new_row = row + dr
        new_col = col + dc

        # Check valid position
        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_position = new_row * 3 + new_col

            # Create new state
            new_state = list(state)

            # Swap blank with tile
            new_state[zero], new_state[new_position] = \
                new_state[new_position], new_state[zero]

            neighbors.append((tuple(new_state), move))

    return neighbors


# Depth Limited Search
def depth_limited_search(state, goal, depth, path, visited):

    print("  " * (3 - depth), "Depth:", depth,
          "Move:", path[-1] if path else "START")

    # Goal test
    if state == goal:
        return path

    # Depth limit reached
    if depth == 0:
        return None

    # Mark state as visited in current path
    visited.add(state)

    # Generate next states
    for next_state, move in get_neighbors(state):

        if next_state not in visited:

            result = depth_limited_search(
                next_state,
                goal,
                depth - 1,
                path + [move],
                visited
            )

            if result is not None:
                return result

    # Remove state for another path
    visited.remove(state)

    return None


# IDS Algorithm
def ids(initial, goal):

    depth = 0

    while True:

        print("\n==============================")
        print("DEPTH LIMIT =", depth)
        print("==============================")

        visited = set()

        result = depth_limited_search(
            initial,
            goal,
            depth,
            [],
            visited
        )

        if result is not None:
            return result

        depth += 1


# Run IDS
print("INITIAL STATE:")
print_puzzle(initial)

print("GOAL STATE:")
print_puzzle(goal)

solution = ids(initial, goal)

print("\n==============================")
print("GOAL FOUND!")
print("==============================")

print("Solution moves:", solution)
print("Number of moves:", len(solution))
```
