```python
import random

# Create an empty board
board = [' ' for _ in range(9)]


# Function to display the board
def display_board():
    print()
    print(" " + board[0] + " | " + board[1] + " | " + board[2])
    print("---+---+---")
    print(" " + board[3] + " | " + board[4] + " | " + board[5])
    print("---+---+---")
    print(" " + board[6] + " | " + board[7] + " | " + board[8])
    print()


# Function to check whether a player has won
def check_winner(symbol):

    winning_positions = [
        (0, 1, 2),  # First row
        (3, 4, 5),  # Second row
        (6, 7, 8),  # Third row
        (0, 3, 6),  # First column
        (1, 4, 7),  # Second column
        (2, 5, 8),  # Third column
        (0, 4, 8),  # Diagonal
        (2, 4, 6)   # Diagonal
    ]

    for a, b, c in winning_positions:
        if board[a] == symbol and board[b] == symbol and board[c] == symbol:
            return True

    return False


# Display position numbers
print("TIC-TAC-TOE")
print()
print("Choose positions using the numbers below:")
print()
print(" 1 | 2 | 3 ")
print("---+---+---")
print(" 4 | 5 | 6 ")
print("---+---+---")
print(" 7 | 8 | 9 ")
print()

print("HUMAN = X")
print("COMPUTER = O")


# Game loop
while True:

    # -------------------------
    # HUMAN TURN
    # -------------------------
    display_board()

    position = int(input("Enter your position (1-9): ")) - 1

    # Check valid position
    if position < 0 or position > 8:
        print("Invalid position! Choose between 1 and 9.")
        continue

    # Check whether cell is already occupied
    if board[position] != ' ':
        print("This position is already occupied!")
        continue

    # Place X
    board[position] = 'X'

    # Check if HUMAN wins
    if check_winner('X'):
        display_board()
        print("🎉 HUMAN WINS!")
        break

    # Check for draw
    if ' ' not in board:
        display_board()
        print("🤝 GAME DRAW!")
        break


    # -------------------------
    # COMPUTER TURN
    # -------------------------

    # Find empty positions
    empty_positions = []

    for i in range(9):
        if board[i] == ' ':
            empty_positions.append(i)

    # Computer randomly selects an empty position
    computer_position = random.choice(empty_positions)

    # Place O
    board[computer_position] = 'O'

    print("Computer chose position:", computer_position + 1)

    # Check if COMPUTER wins
    if check_winner('O'):
        display_board()
        print("💻 COMPUTER WINS!")
        break

    # Check for draw
    if ' ' not in board:
        display_board()
        print("🤝 GAME DRAW!")
        break
```

```python
# Vacuum Cleaner Agent

# Take input from user
room_A = input("Enter status of Room A (Clean/Dirty): ").capitalize()
room_B = input("Enter status of Room B (Clean/Dirty): ").capitalize()
location = input("Enter vacuum location (A/B): ").upper()

print("\nInitial State:")
print("Room A:", room_A)
print("Room B:", room_B)
print("Vacuum Location:", location)

print("\n--- Vacuum Cleaner Starts ---")

# Continue until both rooms are clean
while room_A == "Dirty" or room_B == "Dirty":

    # If vacuum is in Room A
    if location == "A":

        if room_A == "Dirty":
            print("Vacuum is in Room A")
            print("Action: SUCK")
            room_A = "Clean"

        else:
            print("Room A is already clean")
            print("Action: MOVE to Room B")
            location = "B"

    # If vacuum is in Room B
    elif location == "B":

        if room_B == "Dirty":
            print("Vacuum is in Room B")
            print("Action: SUCK")
            room_B = "Clean"

        else:
            print("Room B is already clean")
            print("Action: MOVE to Room A")
            location = "A"

    print("Current State -> Room A:", room_A,
          "| Room B:", room_B,
          "| Vacuum:", location)

# Goal achieved
print("\n--- FINAL STATE ---")
print("Room A:", room_A)
print("Room B:", room_B)
print("Vacuum Location:", location)

print("\n🎯 GOAL ACHIEVED!")
print("Both rooms are clean.")
```
